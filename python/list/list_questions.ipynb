## Q1. How do you create a list in Python?

```python
fruits = ["apple", "banana", "cherry"]
numbers = [1, 2, 3, 4, 5]
mixed = [1, "two", 3.0, True]
empty = []

print(fruits)
print(numbers)
print(mixed)
print(empty)
```

## Q2. How do you access elements using positive and negative indexing?

```python
fruits = ["apple", "banana", "cherry", "date"]

print("First element (index 0):", fruits[0])
print("Third element (index 2):", fruits[2])
print("Last element (index -1):", fruits[-1])
print("Second last element (index -2):", fruits[-2])
```

## Q3. How do you access a range of elements using slicing?

```python
numbers = [10, 20, 30, 40, 50, 60, 70]

print("Elements from index 1 to 3:", numbers[1:4])
print("First three elements:", numbers[:3])
print("From index 3 to end:", numbers[3:])
print("Every second element:", numbers[::2])
print("Reversed list using slicing:", numbers[::-1])
```

## Q4. How do you add a single element to the end of a list using append()?

```python
fruits = ["apple", "banana"]
fruits.append("cherry")
print(fruits)
```

## Q5. How do you insert an element at a specific position using insert()?

```python
fruits = ["apple", "banana", "date"]
fruits.insert(2, "cherry")   # insert "cherry" at index 2
print(fruits)
```

## Q6. How do you add multiple elements to a list using extend()?

```python
numbers = [1, 2, 3]
numbers.extend([4, 5, 6])
print(numbers)

# Note: append() would add the whole list as ONE element instead
demo = [1, 2, 3]
demo.append([4, 5, 6])
print("Using append instead of extend:", demo)
```

## Q7. How do you remove an element by its value using remove()?

```python
fruits = ["apple", "banana", "cherry", "banana"]
fruits.remove("banana")   # removes only the FIRST occurrence
print(fruits)
```

## Q8. How do you remove an element by its index using pop()?

```python
numbers = [10, 20, 30, 40]
removed_item = numbers.pop(1)     # removes element at index 1
print("Removed:", removed_item)
print("Remaining list:", numbers)

last_item = numbers.pop()         # pop() with no index removes the LAST item
print("Last item removed:", last_item)
print("Remaining list:", numbers)
```

## Q9. How do you delete an element or the entire list using del?

```python
numbers = [10, 20, 30, 40, 50]
del numbers[0]           # delete element at index 0
print(numbers)

del numbers[1:3]         # delete a slice of elements
print(numbers)

# del numbers            # this would delete the entire list variable
```

## Q10. How do you empty a list completely using clear()?

```python
numbers = [1, 2, 3, 4, 5]
numbers.clear()
print(numbers)
```

## Q11. How do you find the length of a list using len()?

```python
fruits = ["apple", "banana", "cherry"]
print("Number of elements:", len(fruits))
```

## Q12. How do you check if an element exists in a list using the 'in' keyword?

## Q13. How do you count how many times a value appears using count()?

```python
numbers = [1, 2, 2, 3, 2, 4, 2]
print("Count of 2:", numbers.count(2))
```

## Q14. How do you find the index of the first occurrence of a value using index()?

```python
fruits = ["apple", "banana", "cherry", "banana"]
print("Index of 'banana':", fruits.index("banana"))
```

## Q15. How do you sort a list using sort() and sorted()?

```python
numbers = [5, 2, 8, 1, 9]

# sort() modifies the list in place and returns None
numbers.sort()
print("Sorted in place:", numbers)

numbers.sort(reverse=True)
print("Sorted descending:", numbers)

# sorted() returns a NEW sorted list, original stays unchanged
original = [5, 2, 8, 1, 9]
new_sorted = sorted(original)
print("Original untouched:", original)
print("New sorted list:", new_sorted)
```

## Q16. How do you reverse a list using reverse() and slicing?

```python
numbers = [1, 2, 3, 4, 5]

numbers.reverse()          # reverses the list in place
print("Reversed in place:", numbers)

original = [1, 2, 3, 4, 5]
reversed_copy = original[::-1]   # slicing creates a new reversed list
print("Original untouched:", original)
print("Reversed copy:", reversed_copy)
```

## Q17. How do you copy a list, and why is copy() safer than plain assignment?

```python
original = [1, 2, 3]

# Plain assignment does NOT create a new list, both names point to the same list
reference = original
reference.append(4)
print("Original also changed:", original)

# copy() creates an actual independent copy
original2 = [1, 2, 3]
real_copy = original2.copy()
real_copy.append(99)
print("Original2 untouched:", original2)
print("Copy modified:", real_copy)
```

## Q18. How do you iterate over a list using a for loop?

```python
fruits = ["apple", "banana", "cherry"]

for fruit in fruits:
    print(fruit)

print()
for index, fruit in enumerate(fruits):
    print(f"Index {index}: {fruit}")
```

## Q19. How do you create a list using list comprehension?

```python
squares = [x ** 2 for x in range(1, 6)]
print("Squares:", squares)

even_numbers = [x for x in range(1, 21) if x % 2 == 0]
print("Even numbers:", even_numbers)

upper_fruits = [fruit.upper() for fruit in ["apple", "banana", "cherry"]]
print("Uppercase fruits:", upper_fruits)
```

## Q20. What is a nested list, and how do you access elements inside it?

```python
matrix = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9]
]

print("Whole matrix:", matrix)
print("Second row:", matrix[1])
print("Element at row 2, column 3:", matrix[1][2])

for row in matrix:
    for value in row:
        print(value, end=" ")
    print()
```

## Q21. How do you typecast other data types (string, tuple, set, range) into a list using list()?

```python
# string -> list of characters
word_list = list("hello")
print("String to list:", word_list)

# tuple -> list
tuple_data = (1, 2, 3)
tuple_to_list = list(tuple_data)
print("Tuple to list:", tuple_to_list)

# set -> list
set_data = {10, 20, 30}
set_to_list = list(set_data)
print("Set to list:", set_to_list)

# range -> list
range_to_list = list(range(1, 6))
print("Range to list:", range_to_list)
```

## Q22. How do you concatenate and repeat lists using + and *?

```python
list1 = [1, 2, 3]
list2 = [4, 5, 6]

combined = list1 + list2
print("Concatenated:", combined)

repeated = list1 * 3
print("Repeated:", repeated)
```

## Q23. How do you find the minimum, maximum, and sum of a list?

```python
numbers = [23, 5, 67, 12, 89, 1]

print("Minimum:", min(numbers))
print("Maximum:", max(numbers))
print("Sum:", sum(numbers))
```
