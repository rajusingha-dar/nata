# Nested Dictionary + List — Complex Query Practice Set

Now that you're comfortable with list and dictionary basics, this notebook steps up the difficulty: every question queries a **nested data structure** - a dictionary of employees, where each employee record is itself a dictionary containing a **list** of skills and a **list of dictionaries** for their projects. You'll practice filtering, aggregating, grouping, flattening, sorting, and updating this nested structure - the kind of data-wrangling you'll actually do with real JSON-like data.

## Setup — the dataset used for every question below

Run this cell first. It defines `employee_data`: a dictionary where each **key is an employee ID** and each **value is a dictionary** containing the employee's name, department, salary, a **list** of skills, and a **list of dictionaries** representing their projects. This is the nested dict → list → dict shape you'll query in every question that follows.

```python
employee_data = {
    101: {
        "name": "Raju",
        "department": "Engineering",
        "salary": 85000,
        "skills": ["Python", "SQL", "AWS"],
        "projects": [
            {"name": "Alpha", "status": "completed", "hours": 120},
            {"name": "Beta", "status": "ongoing", "hours": 60},
        ]
    },
    102: {
        "name": "Meera",
        "department": "Engineering",
        "salary": 78000,
        "skills": ["Java", "SQL", "Docker"],
        "projects": [
            {"name": "Gamma", "status": "completed", "hours": 200},
        ]
    },
    103: {
        "name": "Aman",
        "department": "Marketing",
        "salary": 65000,
        "skills": ["SEO", "Content Writing", "Python"],
        "projects": [
            {"name": "Delta", "status": "ongoing", "hours": 40},
            {"name": "Epsilon", "status": "completed", "hours": 90},
            {"name": "Zeta", "status": "ongoing", "hours": 30},
        ]
    },
    104: {
        "name": "Priya",
        "department": "Sales",
        "salary": 72000,
        "skills": ["Negotiation", "CRM", "SQL"],
        "projects": [
            {"name": "Eta", "status": "completed", "hours": 150},
        ]
    },
    105: {
        "name": "Karan",
        "department": "Engineering",
        "salary": 95000,
        "skills": ["Python", "AWS", "Kubernetes", "Docker"],
        "projects": [
            {"name": "Theta", "status": "ongoing", "hours": 80},
            {"name": "Iota", "status": "completed", "hours": 110},
        ]
    },
}

print(f"Loaded {len(employee_data)} employee records.")
```

## Q1. Access a deeply nested value: print Raju's (ID 101) second skill in the list.

```python
second_skill = employee_data[101]["skills"][1]
print("Raju's second skill:", second_skill)
```

## Q2. Print all project names for Aman (ID 103) using a list comprehension over his nested project list.

```python
aman_project_names = [project["name"] for project in employee_data[103]["projects"]]
print("Aman's projects:", aman_project_names)
```

## Q3. Find the names of every employee who knows a given skill (e.g. 'Python'), by searching inside the nested skills list of every employee.

```python
target_skill = "Python"
python_devs = [emp["name"] for emp in employee_data.values() if target_skill in emp["skills"]]
print(f"Employees who know {target_skill}:", python_devs)
```

## Q4. Build a new dictionary mapping each employee's name to their TOTAL project hours (summing across the nested list of project dicts).

```python
total_hours_by_name = {
    emp["name"]: sum(project["hours"] for project in emp["projects"])
    for emp in employee_data.values()
}
print(total_hours_by_name)
```

## Q5. Using the result from the previous question's logic, find the single employee with the MAXIMUM total project hours.

```python
total_hours_by_name = {
    emp["name"]: sum(project["hours"] for project in emp["projects"])
    for emp in employee_data.values()
}
top_employee = max(total_hours_by_name, key=total_hours_by_name.get)
print(f"Employee with most hours: {top_employee} ({total_hours_by_name[top_employee]} hours)")
```

## Q6. Get a sorted list of every UNIQUE skill across the whole company (flatten all the nested skills lists into one set, then sort it).

```python
all_skills = {skill for emp in employee_data.values() for skill in emp["skills"]}
print("All unique skills:", sorted(all_skills))
```

## Q7. Group employees by department: build a dictionary mapping department name -> list of employee names in that department.

```python
department_map = {}
for emp in employee_data.values():
    dept = emp["department"]
    department_map.setdefault(dept, []).append(emp["name"])

print(department_map)
```

## Q8. Calculate the AVERAGE salary per department (department -> average salary).

```python
dept_totals = {}
dept_counts = {}

for emp in employee_data.values():
    dept = emp["department"]
    dept_totals[dept] = dept_totals.get(dept, 0) + emp["salary"]
    dept_counts[dept] = dept_counts.get(dept, 0) + 1

dept_average_salary = {
    dept: dept_totals[dept] / dept_counts[dept] for dept in dept_totals
}
print(dept_average_salary)
```

## Q9. List the names of employees who currently have AT LEAST ONE 'ongoing' project.

```python
employees_with_ongoing = [
    emp["name"]
    for emp in employee_data.values()
    if any(project["status"] == "ongoing" for project in emp["projects"])
]
print("Have an ongoing project:", employees_with_ongoing)
```

## Q10. Sort all employees by salary in DESCENDING order and print each as 'Name - Salary'.

```python
sorted_by_salary = sorted(employee_data.values(), key=lambda emp: emp["salary"], reverse=True)

for emp in sorted_by_salary:
    print(f"{emp['name']} - {emp['salary']}")
```

## Q11. Flatten every project across every employee and count how many are 'completed' vs 'ongoing' overall.

```python
status_counts = {}

for emp in employee_data.values():
    for project in emp["projects"]:
        status = project["status"]
        status_counts[status] = status_counts.get(status, 0) + 1

print(status_counts)
```

## Q12. Find the names of all projects (across every employee) where the hours worked is greater than 100.

```python
high_effort_projects = [
    project["name"]
    for emp in employee_data.values()
    for project in emp["projects"]
    if project["hours"] > 100
]
print("Projects with more than 100 hours:", high_effort_projects)
```

## Q13. Give a 10% salary raise to every employee who has completed MORE THAN ONE project, then print each employee's updated salary. (This mutates the nested dictionary in place.)

```python
for emp in employee_data.values():
    completed_count = sum(1 for project in emp["projects"] if project["status"] == "completed")
    if completed_count > 1:
        emp["salary"] = round(emp["salary"] * 1.10, 2)

for emp in employee_data.values():
    print(f"{emp['name']}: {emp['salary']}")
```

## Q14. Flatten the ENTIRE nested database into a list of flat dictionaries - one record per project - with keys: employee_id, employee_name, department, project_name, status, hours. (Useful for exporting to a table or CSV.)

```python
flat_records = [
    {
        "employee_id": emp_id,
        "employee_name": emp["name"],
        "department": emp["department"],
        "project_name": project["name"],
        "status": project["status"],
        "hours": project["hours"],
    }
    for emp_id, emp in employee_data.items()
    for project in emp["projects"]
]

for record in flat_records:
    print(record)
```

## Q15. Find which department has the HIGHEST total salary expense (sum of salaries of every employee in that department).

```python
dept_salary_totals = {}
for emp in employee_data.values():
    dept = emp["department"]
    dept_salary_totals[dept] = dept_salary_totals.get(dept, 0) + emp["salary"]

top_department = max(dept_salary_totals, key=dept_salary_totals.get)
print("Department-wise salary totals:", dept_salary_totals)
print(f"Highest total salary expense: {top_department} ({dept_salary_totals[top_department]})")
```
