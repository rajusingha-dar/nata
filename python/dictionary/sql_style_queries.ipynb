# SQL-Style Queries in Pure Python — Nested Dictionary Practice Set

This notebook takes SQL-style questions (`SELECT`, `WHERE`, `GROUP BY`, `ORDER BY`, `HAVING`, `JOIN`, `LIMIT`, `DISTINCT`, `BETWEEN`) and asks you to answer them using **plain Python** over nested dictionaries - no SQL, no pandas. The dataset models two tables (`orders` and `customers`) the way you'd often see JSON/API data structured in the real world: a dictionary keyed by ID, with each record being another dictionary.

## Setup — the two "tables" used for every question below

Run this cell first. `orders` is a dictionary keyed by `order_id`, where each value is a dictionary of order details (this is your **Orders table**). `customers` is a dictionary keyed by `customer_id` (this is your **Customers table**). Together they let you practice SQL-style filtering, grouping, sorting, and joins - written as plain Python.

```python
orders = {
    1:  {"customer_id": 101, "product": "Laptop",     "category": "Electronics", "quantity": 1, "price": 55000, "city": "Kolkata",   "order_date": "2024-01-15"},
    2:  {"customer_id": 102, "product": "Mouse",       "category": "Electronics", "quantity": 3, "price": 500,   "city": "Delhi",     "order_date": "2024-01-20"},
    3:  {"customer_id": 101, "product": "Desk",        "category": "Furniture",   "quantity": 1, "price": 8000,  "city": "Kolkata",   "order_date": "2024-02-05"},
    4:  {"customer_id": 103, "product": "Chair",       "category": "Furniture",   "quantity": 4, "price": 2000,  "city": "Mumbai",    "order_date": "2024-02-10"},
    5:  {"customer_id": 102, "product": "Keyboard",    "category": "Electronics", "quantity": 2, "price": 1200,  "city": "Delhi",     "order_date": "2024-02-15"},
    6:  {"customer_id": 104, "product": "Monitor",     "category": "Electronics", "quantity": 1, "price": 12000, "city": "Bangalore", "order_date": "2024-03-01"},
    7:  {"customer_id": 101, "product": "Bookshelf",   "category": "Furniture",   "quantity": 2, "price": 3500,  "city": "Kolkata",   "order_date": "2024-03-10"},
    8:  {"customer_id": 103, "product": "Lamp",        "category": "Furniture",   "quantity": 3, "price": 800,   "city": "Mumbai",    "order_date": "2024-03-15"},
    9:  {"customer_id": 105, "product": "Headphones",  "category": "Electronics", "quantity": 2, "price": 2500,  "city": "Chennai",   "order_date": "2024-04-01"},
    10: {"customer_id": 102, "product": "Table",       "category": "Furniture",   "quantity": 1, "price": 6000,  "city": "Delhi",     "order_date": "2024-04-05"},
}

customers = {
    101: {"name": "Raju",  "email": "raju@example.com"},
    102: {"name": "Meera", "email": "meera@example.com"},
    103: {"name": "Aman",  "email": "aman@example.com"},
    104: {"name": "Priya", "email": "priya@example.com"},
    105: {"name": "Karan", "email": "karan@example.com"},
}

print(f"Loaded {len(orders)} orders and {len(customers)} customers.")
```

## Q1. `SELECT * FROM orders WHERE category = 'Electronics';`

```python
electronics_orders = [
    {"order_id": oid, **order}
    for oid, order in orders.items()
    if order["category"] == "Electronics"
]

for order in electronics_orders:
    print(order)
```

## Q2. `SELECT DISTINCT city FROM orders;`

```python
distinct_cities = sorted({order["city"] for order in orders.values()})
print(distinct_cities)
```

## Q3. `SELECT customer_id, SUM(price * quantity) AS total_spent FROM orders GROUP BY customer_id ORDER BY total_spent DESC;`

```python
spend_by_customer = {}
for order in orders.values():
    cust_id = order["customer_id"]
    order_value = order["price"] * order["quantity"]
    spend_by_customer[cust_id] = spend_by_customer.get(cust_id, 0) + order_value

# ORDER BY total_spent DESC
sorted_spend = sorted(spend_by_customer.items(), key=lambda item: item[1], reverse=True)

for cust_id, total_spent in sorted_spend:
    print(f"customer_id={cust_id}, total_spent={total_spent}")
```

## Q4. `SELECT category, COUNT(*) AS order_count FROM orders GROUP BY category;`

```python
order_count_by_category = {}
for order in orders.values():
    category = order["category"]
    order_count_by_category[category] = order_count_by_category.get(category, 0) + 1

print(order_count_by_category)
```

## Q5. `SELECT * FROM orders WHERE quantity > 2 ORDER BY price DESC;`

```python
filtered = [
    {"order_id": oid, **order}
    for oid, order in orders.items()
    if order["quantity"] > 2
]

filtered_sorted = sorted(filtered, key=lambda order: order["price"], reverse=True)

for order in filtered_sorted:
    print(order)
```

## Q6. `SELECT category, AVG(price) AS avg_price FROM orders GROUP BY category HAVING avg_price > 3000;`

```python
price_sum_by_category = {}
count_by_category = {}

for order in orders.values():
    category = order["category"]
    price_sum_by_category[category] = price_sum_by_category.get(category, 0) + order["price"]
    count_by_category[category] = count_by_category.get(category, 0) + 1

avg_price_by_category = {
    category: price_sum_by_category[category] / count_by_category[category]
    for category in price_sum_by_category
}

# HAVING avg_price > 3000
result = {
    category: avg_price
    for category, avg_price in avg_price_by_category.items()
    if avg_price > 3000
}

print(result)
```

## Q7. `SELECT o.product, c.name, c.email FROM orders o JOIN customers c ON o.customer_id = c.id;`

```python
joined_result = []

for order in orders.values():
    customer = customers[order["customer_id"]]   # lookup = the "join"
    joined_result.append({
        "product": order["product"],
        "name": customer["name"],
        "email": customer["email"],
    })

for row in joined_result:
    print(row)
```

## Q8. `SELECT * FROM orders ORDER BY (price * quantity) DESC LIMIT 3;`

```python
orders_with_total = [
    {"order_id": oid, **order, "order_value": order["price"] * order["quantity"]}
    for oid, order in orders.items()
]

top_3 = sorted(orders_with_total, key=lambda order: order["order_value"], reverse=True)[:3]

for order in top_3:
    print(order)
```

## Q9. `SELECT city, COUNT(DISTINCT customer_id) AS unique_customers FROM orders GROUP BY city;`

```python
customers_by_city = {}
for order in orders.values():
    city = order["city"]
    customers_by_city.setdefault(city, set()).add(order["customer_id"])

unique_customers_by_city = {
    city: len(customer_ids) for city, customer_ids in customers_by_city.items()
}

print(unique_customers_by_city)
```

## Q10. `SELECT * FROM orders WHERE order_date BETWEEN '2024-02-01' AND '2024-03-31';`

```python
start_date = "2024-02-01"
end_date = "2024-03-31"

orders_in_range = [
    {"order_id": oid, **order}
    for oid, order in orders.items()
    if start_date <= order["order_date"] <= end_date
]

for order in orders_in_range:
    print(order)
```
