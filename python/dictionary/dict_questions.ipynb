# Python Dictionary — Question & Answer Practice Set

This notebook walks through the fundamentals of Python dictionaries as a set of questions, each followed by a short runnable example — access, get(), add/update, remove (pop/popitem/del/clear), keys/values/items, nested dictionaries, dictionary comprehension, merging, sorting, and typecasting to and from lists.

## Q1. How do you create a dictionary in Python?

```python
student = {"name": "Raju", "age": 30, "city": "Kolkata"}
empty_dict = {}
using_dict_func = dict(name="Meera", age=25)

print(student)
print(empty_dict)
print(using_dict_func)
```

## Q2. How do you access a value directly using its key?

```python
student = {"name": "Raju", "age": 30, "city": "Kolkata"}

print(student["name"])
print(student["age"])

# Accessing a missing key directly raises a KeyError:
# print(student["marks"])   # this would crash the program
```

## Q3. How do you access a value safely using get() (without crashing on a missing key)?

```python
student = {"name": "Raju", "age": 30}

print(student.get("name"))
print(student.get("marks"))            # returns None if key is missing
print(student.get("marks", "Not Found"))   # returns a default value instead
```

## Q4. How do you add a new key-value pair to a dictionary?

```python
student = {"name": "Raju", "age": 30}
student["city"] = "Kolkata"     # adds a new key
print(student)
```

## Q5. How do you update an existing value in a dictionary?

```python
student = {"name": "Raju", "age": 30}
student["age"] = 31             # updates existing key
print(student)

# update() can change multiple keys at once
student.update({"age": 32, "city": "Kolkata"})
print(student)
```

## Q6. How do you remove a key-value pair using pop()?

```python
student = {"name": "Raju", "age": 30, "city": "Kolkata"}

removed_value = student.pop("city")
print("Removed value:", removed_value)
print("Remaining dictionary:", student)

# pop() with a default avoids a KeyError if the key does not exist
missing = student.pop("marks", "Key not found")
print(missing)
```

## Q7. How do you remove the last inserted key-value pair using popitem()?

```python
student = {"name": "Raju", "age": 30, "city": "Kolkata"}

last_item = student.popitem()
print("Removed item:", last_item)
print("Remaining dictionary:", student)
```

## Q8. How do you delete a specific key using del?

```python
student = {"name": "Raju", "age": 30, "city": "Kolkata"}
del student["city"]
print(student)

# del student           # this would delete the entire dictionary variable
```

## Q9. How do you empty a dictionary completely using clear()?

```python
student = {"name": "Raju", "age": 30}
student.clear()
print(student)
```

## Q10. How do you get all the keys of a dictionary using keys()?

```python
student = {"name": "Raju", "age": 30, "city": "Kolkata"}
print(student.keys())
```

## Q11. How do you get all the values of a dictionary using values()?

```python
student = {"name": "Raju", "age": 30, "city": "Kolkata"}
print(student.values())
```

## Q12. How do you get all key-value pairs using items()?

```python
student = {"name": "Raju", "age": 30, "city": "Kolkata"}
print(student.items())

for key, value in student.items():
    print(f"{key} -> {value}")
```

## Q13. How do you check if a key exists in a dictionary using 'in'?

```python
student = {"name": "Raju", "age": 30}

print("name" in student)
print("marks" in student)
print("marks" not in student)
```

## Q14. How do you iterate over a dictionary's keys, values, and items?

```python
student = {"name": "Raju", "age": 30, "city": "Kolkata"}

print("Iterating over keys:")
for key in student:
    print(key)

print("\nIterating over values:")
for value in student.values():
    print(value)

print("\nIterating over key-value pairs:")
for key, value in student.items():
    print(key, "=", value)
```

## Q15. How do you find the number of key-value pairs using len()?

```python
student = {"name": "Raju", "age": 30, "city": "Kolkata"}
print("Number of items:", len(student))
```

## Q16. How do you copy a dictionary safely using copy()?

```python
original = {"name": "Raju", "age": 30}

# Plain assignment just creates another reference to the SAME dictionary
reference = original
reference["age"] = 99
print("Original also changed:", original)

# copy() creates an independent copy
original2 = {"name": "Meera", "age": 25}
real_copy = original2.copy()
real_copy["age"] = 50
print("Original2 untouched:", original2)
print("Copy modified:", real_copy)
```

## Q17. How do you merge two dictionaries using update() and the | operator?

```python
dict1 = {"a": 1, "b": 2}
dict2 = {"b": 20, "c": 3}

merged_with_update = dict1.copy()
merged_with_update.update(dict2)
print("Merged with update():", merged_with_update)

merged_with_pipe = dict1 | dict2   # Python 3.9+
print("Merged with | operator:", merged_with_pipe)
```

## Q18. How do you create a dictionary using dictionary comprehension?

```python
squares = {x: x ** 2 for x in range(1, 6)}
print("Squares dictionary:", squares)

names = ["apple", "banana", "cherry"]
name_lengths = {name: len(name) for name in names}
print("Name lengths:", name_lengths)
```

## Q19. What is a nested dictionary, and how do you access values inside it?

```python
students = {
    1: {"name": "Raju", "marks": {"Math": 90, "Science": 85}},
    2: {"name": "Meera", "marks": {"Math": 60, "Science": 70}}
}

print("All data:", students)
print("Student 1's name:", students[1]["name"])
print("Student 1's Math marks:", students[1]["marks"]["Math"])

for student_id, info in students.items():
    print(f"ID {student_id}: {info['name']} -> {info['marks']}")
```

## Q20. How do you typecast a dictionary's keys, values, and items into lists?

```python
student = {"name": "Raju", "age": 30, "city": "Kolkata"}

keys_list = list(student.keys())
values_list = list(student.values())
items_list = list(student.items())

print("Keys as list:", keys_list)
print("Values as list:", values_list)
print("Items as list of tuples:", items_list)
```

## Q21. How do you typecast a list of tuples (or two lists) into a dictionary using dict()?

```python
pairs = [("name", "Raju"), ("age", 30), ("city", "Kolkata")]
from_pairs = dict(pairs)
print("Dictionary from list of tuples:", from_pairs)

keys = ["name", "age", "city"]
values = ["Meera", 25, "Delhi"]
from_zip = dict(zip(keys, values))
print("Dictionary from two lists using zip():", from_zip)
```

## Q22. How do you use setdefault() to get a value or insert a default if the key is missing?

```python
student = {"name": "Raju", "age": 30}

age = student.setdefault("age", 0)         # key exists, returns existing value
print("Age:", age)

city = student.setdefault("city", "Unknown")  # key missing, adds it with default
print("City:", city)
print("Dictionary after setdefault:", student)
```

## Q23. How do you sort a dictionary by its keys or by its values?

```python
marks = {"Meera": 85, "Raju": 92, "Aman": 70}

sorted_by_key = dict(sorted(marks.items()))
print("Sorted by key:", sorted_by_key)

sorted_by_value = dict(sorted(marks.items(), key=lambda pair: pair[1], reverse=True))
print("Sorted by value (highest first):", sorted_by_value)
```

## Q24. How do you check if a dictionary is empty?

```python
empty_dict = {}
filled_dict = {"name": "Raju"}

if not empty_dict:
    print("empty_dict is empty")

if filled_dict:
    print("filled_dict has data")
```
